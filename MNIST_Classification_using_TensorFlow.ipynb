```python
import tensorflow as tf
```

```python
#هنا نحمل مجموعة بيانات MNIST، وهي صور لأرقام مكتوبة يدويًا من 0 إلى 9.
(x_train,y_train),(x_test, y_test) =tf.keras.datasets.mnist.load_data()

#تطبيع البيانات: نحول قيم البكسل من 0–255 إلى 0–1 لتسهيل عملية التدريب وتحسين دقة النموذج.
x_train=x_train/255.0
x_test=x_test /255.0
```

```python
#هنا نبني نموذج شبكة عصبية بسيطة (MLP) باستخدام Sequential
model=tf.keras.Sequential([
    tf.keras.layers.Flatten(input_shape=(28,28)),#تحويل الصورة إلى بعد واحد
    tf.keras.layers.Dense(128,activation='relu'),#استخدام دالة التفعيل
    tf.keras.layers.Dense(10,activation='softmax')#تحويل الناتج إلى احتمالات
])
```

```python
#تجميع النموذج: نحدد طريقة التدريب.
model.compile(optimizer='adam',#خوارزمية تحسين شائعة لتحديث الأوزان بسرعة وكفاءة
              loss='sparse_categorical_crossentropy',#دالة الفقد (loss) المناسبة إذا كانت الملصقات أرقام صحيحة.
              metrics=['accuracy'])#لمتابعة دقة النموذج أثناء التدريب.
```

```python
#تدريب النموذج على بيانات التدريب
model.fit(x_train,y_train,
          epochs=1,#عدد مرات مرور كامل على مجموعة البيانات
          batch_size=64) #تقسيم البيانات إلى دفعات من 64 صورة لتسريع التدريب
```

```python
#اختبار النموذج على بيانات جديدة
test_loss, test_acc = model.evaluate(x_test, y_test)#evaluate يعيد الخسارة والدقة على بيانات لم يشاهدها النموذج.
print("دقة النموذج على بيانات الاختبار:", test_acc)#test_acc هي نسبة التوقعات الصحيحة على مجموعة الاختبار.
```

```python
#لرؤية النتيجة على عينة من بيانات الاختبار
predictions = model.predict(x_test[:5])#هنا نأخذ أول 5 صور من مجموعة الاختبار
#الناتج predictions سيكون مصفوفة 2D
for i, pred in enumerate(predictions):#enumerate تعطيني الفهرس والبيانات معًا
    print(f"الصورة {i} , النموذج توقع: {pred.argmax()}, القيمة الحقيقية: {y_test[i]}")
```
